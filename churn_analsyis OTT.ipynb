```python
#AVERAGE REVENUE PER USER
arpu=df['monthly_charges'].mean()
print("ARPU  = ",round(arpu,2))
```

```python
#AVERAGE CUSTOMER TENURE
#COUNT OF DAY USER HAS USED FOR OUR SERVICE: CANCELLATION DATE ELSE CURRENT DATE 
today=pd.Timestamp.today()
df['tenure_days']=np.where(
       df['cancellation_date'].notna(),
       (df['cancellation_date'] - df['subscription_start_date']).dt.days,
    (today - df['subscription_start_date']).dt.days
   )
avg_ten=df['tenure_days'].mean()
print("Average tenure =" ,round(avg_ten,0))
```

```python
#10. CORREALTION ESCALATION VS CHURN
import warnings

# Filter out standard NumPy RuntimeWarnings
warnings.filterwarnings("ignore", category=RuntimeWarning)

df['escalations']=np.where(df['escalations'] == 'Y',1,0)

corr_df=df[['escalations','churn_flag']].dropna()
#correlation =
correlation=corr_df['escalations'].corr(df['churn_flag'])
print("Correlation between escalation and churn is =",round(correlation,2))
```
